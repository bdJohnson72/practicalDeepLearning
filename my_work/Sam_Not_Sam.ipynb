# Sam or Not Sam
This notebook is to learn the basic ideas of chapter 1 of the FastAI course and
implement a basic deep learning tool to recognize is a beer is Sam Adams or not Sam Adams


```python
from fastai.vision.all import *
```

```python
# get the Duck Duck Go Search tools
from duckduckgo_search import DDGS
from fastcore.all import *

def search_images(keywords, max_images=200):
    return L(DDGS().images(keywords, max_results=max_images)).itemgot('image')

# It's extracting just the 'image' URLs from each result dictionary!
```

```python
sam_images = search_images(keywords='Samuel Adams Beer', max_images=200)


from fastdownload import download_url
dest = 'sam.jpg'
download_url(sam_images[0], dest, show_progress=False)

from fastai.vision.all import *
im = Image.open(dest)
im.to_thumb(256,256)
```

```python
other_beers = search_images(keywords='beer bottle', max_images=1000)
dest = 'other.jpg'
download_url(other_beers[5], dest, show_progress=False)
other = Image.open(dest)
other.to_thumb(256,256)
```

```python
canned_beers = search_images(keywords='beer can', max_images=1000)
dest = 'canned.jpg'
download_url(canned_beers[5], dest, show_progress=False)
canned = Image.open(dest)
canned.to_thumb(256,256)
```

```python
other_beer_urls = []
for beer in other_beers:
    other_beer_urls.itemgot(beer.url)

print(len(other_beer_urls))
```
